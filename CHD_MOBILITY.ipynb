```python
import pandas as pd
import numpy as np
import folium
import random

# 1. SIMULATE: Generate 5,000 fake commuter survey responses for Chandigarh
np.random.seed(42)
data = {
    'Survey_ID': range(1, 5001),
    'Transport_Mode': np.random.choice(['Car', 'Two-Wheeler', 'Cycle', 'CTU Bus'], size=5000, p=[0.4, 0.3, 0.1, 0.2]),
    # Generate random GPS points roughly around Chandigarh sectors
    'Latitude': np.random.uniform(30.70, 30.77, size=5000),
    'Longitude': np.random.uniform(76.73, 76.81, size=5000),
    'Satisfaction_Score': np.random.choice([1, 2, 3, 4, 5, np.nan], size=5000, p=[0.1, 0.2, 0.3, 0.2, 0.1, 0.1]) # adds some missing data
}
df = pd.DataFrame(data)

# 2. CLEAN: (Claim 1 practiced!) Clean missing responses in seconds
df['Satisfaction_Score'] = df['Satisfaction_Score'].fillna(df['Satisfaction_Score'].median())

# 3. ANALYZE: Count preferred transport modes instantly
print("Commuter Distribution:\n", df['Transport_Mode'].value_counts())

# 4. MAP: (Claim 2 practiced!) Map the first 100 heavily congested Car commuter points
m = folium.Map(location=[30.7333, 76.7794], zoom_start=13, tiles="OpenStreetMap")
car_commuters = df[df['Transport_Mode'] == 'Car'].head(100)

for idx, row in car_commuters.iterrows():
    folium.CircleMarker(
        location=[row['Latitude'], row['Longitude']],
        radius=4,
        color='red',
        fill=True,
        popup=f"ID: {row['Survey_ID']} - Car User"
    ).add_to(m)

# Save the map as an interactive HTML file
m.save("chandigarh_mobility_map.html")
print("Successfully generated interactive map!")
```
